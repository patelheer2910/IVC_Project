```python
import os
from utils import *
import matplotlib.pyplot as plt

output_data = 'output_unmask_test'
os.makedirs(output_data, exist_ok=True)
for f in os.listdir(input_data):
    # os.makedirs(os.path.join(output_data, direc), exist_ok=True)
    # direc_path = os.path.join(output_data, direc)
    # for f in os.listdir(os.path.join(input_data, direc)):
    try:
        grp = bb_grouped.get_group(f)
        # img = cv2.imread(os.path.join(os.path.join(input_data, direc),f))
        img = cv2.imread(os.path.join(input_data,f))
        img, _ = smart_resize(img, new_size=640)
        mask = np.zeros(img.shape[:2], dtype=np.uint8)
        xmin = list(grp['x1'])
        xmax = list(grp['x2'])
        ymin = list(grp['y1'])
        ymax = list(grp['y2'])
        for i in range(len(xmin)):
            cv2.rectangle(mask, (xmin[i], ymin[i]), (xmax[i], ymax[i]), 255, thickness=-5)
        result = cv2.bitwise_and(img, img, mask=mask)
        cv2.imwrite(os.path.join(output_data,f.replace('.ppm','.jpg')),result)
    except:
        print(f)
```
